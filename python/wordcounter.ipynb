Reading via for line in f splits on newline and it's efficient memory-wise (it reads one line at a time) but putting everything in a huge list is not. Anyway, if you insist:

```python
huge_list = []
""
with open('wordcounter.txt', "r") as f:
    for line in f:
        huge_list.extend(line.split())

print(huge_list)
```

To read the whole file into memory as a string, use f.read() instead:



```python
huge_list = []

with open('wordcounter.txt', "r") as f:
    huge_list = f.read().split()

print(huge_list)
```

```python

stripped_list = [s.strip(',') for s in [s.strip('.') for s in huge_list]] # remove the , & . from the string borders
# print([s.replace('8', '') for s in huge_list]) # remove all the 8s 
```

```python
print(stripped_list)
```

```python
from collections import Counter

element_counts = Counter(stripped_list)

top_10 = element_counts.most_common(10)
print(top_10)
```

```python
lx = [l[0] for l in top_10]
print(lx)
ly = [l[1] for l in top_10]
print(ly)
```

```python

import matplotlib.pyplot as plt
```

```python
plt.barh(lx, ly)
plt.show()
```
